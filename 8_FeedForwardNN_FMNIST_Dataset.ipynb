# 8. Feed Forward Neural Network - Fashion-MNIST Dataset
Uses TensorFlow/Keras. The Fashion-MNIST dataset is auto-downloaded on first run.

```python
import tensorflow as tf
from tensorflow.keras.datasets import fashion_mnist
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Flatten
```

```python
(X_train, y_train), (X_test, y_test) = fashion_mnist.load_data()

X_train = X_train / 255.0
X_test = X_test / 255.0
```

```python
model = Sequential([
    Flatten(input_shape=(28, 28)),
    Dense(128, activation="relu"),
    Dense(64, activation="relu"),
    Dense(10, activation="softmax")
])

model.compile(optimizer="adam", loss="sparse_categorical_crossentropy", metrics=["accuracy"])
model.summary()
```

```python
history = model.fit(X_train, y_train, epochs=5, validation_split=0.1, batch_size=32)
```

```python
test_loss, test_acc = model.evaluate(X_test, y_test)
print("Test accuracy:", test_acc)
```
